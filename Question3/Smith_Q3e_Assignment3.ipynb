```python
import numpy as np
def pixel_coordinates(W,H,n,f,fx,fy,ox,oy,Pc):
  Mproj=[[2*fx/W, 0, (W-2*ox)/W, 0], [0, 2*fy/H, (-H+2*oy)/H, 0], [0, 0, -(f+n)/(f-n), (-2*f*n)/(f-n)], [0, 0, -1, 0] ]
  Pc=np.transpose(Pc)
  Pclip=Mproj@Pc
  PNDC=[Pclip[0]/Pclip[3], Pclip[1]/Pclip[3], Pclip[2]/Pclip[3], 1]
  u=.5*(PNDC[0]+1)*W
  v=.5*(PNDC[1]+1)*H
  return (u,v)
```

```python
pixel_coordinates(800, 600, .2, 20, 800, 800, 400, 400, [.3,-.15,-6,1])
```
